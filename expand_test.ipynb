```python
import torch



samples_dt_20 = torch.load('./nullsamples/amazon20_1024/20/samples_dt.pt')
samples_type_20 = torch.load('./nullsamples/amazon20_1024/20/samples_type.pt')
samples_dt_40 = torch.load('./nullsamples/amazon20_1024/40/samples_dt.pt')
samples_type_40 = torch.load('./nullsamples/amazon20_1024/40/samples_type.pt')
samples_dt_60 = torch.load('./nullsamples/amazon20_1024/60/samples_dt.pt')
samples_type_60 = torch.load('./nullsamples/amazon20_1024/60/samples_type.pt')
samples_dt_80 = torch.load('./nullsamples/amazon20_1024/80/samples_dt.pt')
samples_type_80 = torch.load('./nullsamples/amazon20_1024/80/samples_type.pt')
samples_dt_100 = torch.load('./nullsamples/amazon20_1024/100/samples_dt.pt')
samples_type_100 = torch.load('./nullsamples/amazon20_1024/100/samples_type.pt')
samples_dt_120 = torch.load('./nullsamples/amazon20_1024/120/samples_dt.pt')
samples_type_120 = torch.load('./nullsamples/amazon20_1024/120/samples_type.pt')
samples_dt_140 = torch.load('./nullsamples/amazon20_1024/140/samples_dt.pt')
samples_type_140 = torch.load('./nullsamples/amazon20_1024/140/samples_type.pt')
samples_dt_160 = torch.load('./nullsamples/amazon20_1024/160/samples_dt.pt')
samples_type_160 = torch.load('./nullsamples/amazon20_1024/160/samples_type.pt')
samples_dt_180 = torch.load('./nullsamples/amazon20_1024/180/samples_dt.pt')
samples_type_180 = torch.load('./nullsamples/amazon20_1024/180/samples_type.pt')
samples_dt_200 = torch.load('./nullsamples/amazon20_1024/200/samples_dt.pt')
samples_type_200 = torch.load('./nullsamples/amazon20_1024/200/samples_type.pt')



N = samples_dt_20.size(0)          # Batch size
sequences = []
# print(N)
for i in range(N):
    dt_part1  = samples_dt_20[i, :, 0].tolist()   # 20
    typ_part1 = samples_type_20[i, :, 0].tolist()

    dt_part2  = samples_dt_40[i, :, 0].tolist()   # 20
    typ_part2 = samples_type_40[i, :, 0].tolist()

    dt_part3  = samples_dt_60[i, :, 0].tolist()   # 20
    typ_part3 = samples_type_60[i, :, 0].tolist()

    dt_part4  = samples_dt_80[i, :, 0].tolist()   # 20
    typ_part4 = samples_type_80[i, :, 0].tolist()

    dt_part5  = samples_dt_100[i, :, 0].tolist()   # 20
    typ_part5 = samples_type_100[i, :, 0].tolist()

    dt_part6  = samples_dt_120[i, :, 0].tolist()   # 20
    typ_part6 = samples_type_120[i, :, 0].tolist()

    dt_part7  = samples_dt_140[i, :, 0].tolist()   # 20
    typ_part7 = samples_type_140[i, :, 0].tolist()

    dt_part8  = samples_dt_160[i, :, 0].tolist()   # 20
    typ_part8 = samples_type_160[i, :, 0].tolist()

    dt_part9  = samples_dt_180[i, :, 0].tolist()   # 20
    typ_part9 = samples_type_180[i, :, 0].tolist()

    dt_part10  = samples_dt_200[i, :, 0].tolist()   # 20
    typ_part10 = samples_type_200[i, :, 0].tolist()

    dt_all  = dt_part1 + dt_part2 + dt_part3 + dt_part4 +  dt_part5 +  dt_part6 +  dt_part7 +  dt_part8 +  dt_part9  +  dt_part10
    typ_all = typ_part1 + typ_part2 + typ_part3 +  typ_part4  +   typ_part5 +  typ_part6  +  typ_part7  +  typ_part8 +  typ_part9 +  typ_part10

    seq, t_cum = [], 0.0
    for d, k in zip(dt_all, typ_all):
        t_cum += d
        seq.append({
            'type_event': int(k),
            'time_since_last_event': float(d),
            'time_since_start': t_cum
        })

    # pad = 20                                      
    # seq.extend([{
    #     'type_event': 0,
    #     'time_since_last_event': 0,
    #     'time_since_start': 0
    # } for _ in range(pad)])

    sequences.append(seq)                    

dataset = {
    'dim_process': 1,
    'generated': sequences
}

with open('generated_amazon.pkl', 'wb') as f:
    pickle.dump(dataset, f)
```
